```python
import pandas as pd

students = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Paris', 'London', 'Tokyo']
}
df = pd.DataFrame(students)
df
```

```python
# Index based selection with iloc
print(df.iloc[0:2]) # Rows 0 to 1 can be accessed using the iloc method with a slice.

print(df.iloc[:,0]) # All rows of the first column can be accessed using iloc with a slice.

print(df.iloc[0:2,0:2]) # Rows 0 to 2 and columns 0 to 2 can be accessed using iloc with slices.

print(df.loc[0]) # The first row can be accessed using loc with the row index.
```

```python
# Label Based Selection with loc

print(df.loc[0,'City']) # The value in the first row and the 'City' column can be accessed using loc with the row index and column label.

print(df.loc[:,['Name','City']]) # All rows and the 'Name' and 'City' columns can be accessed using loc with a slice and a list of column labels.

print(df.loc[0:2,['City']]) # Rows 0 to 2 and the 'City' column can be accessed using loc with slices and a list of column labels.
```

```python
# difference between loc and iloc is that loc uses labels to select rows and columns, while iloc uses integer positions.
# when to loc and iloc is used depends on the data and the specific use case. 
# If you want to select rows and columns based on their labels, use loc. 
# If you want to select rows and columns based on their integer positions, use iloc.
```
