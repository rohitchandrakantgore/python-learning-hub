# Pandas Introduction
Pandas is a Python library for working with tabular data in Python. It helps you read, clean, filter, and analyze data efficiently.

## Why use pandas?
- Read CSV and Excel data
- Clean missing or messy values
- Filter and group data
- Perform basic analysis

## Quick concepts
- Series: a one-dimensional labeled array
- DataFrame: a two-dimensional table with rows and columns
- Index: labels for rows or values


````python
import pandas as pd

# Series example
scores = pd.Series([85, 90, 78], index=['Math', 'Science', 'English'])
print('Series:')
print(scores)

# DataFrame example
students = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Age': [25, 30]
})
print('\nDataFrame:')
print(students)
````
