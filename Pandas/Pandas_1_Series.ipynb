# Pandas Series
A Series is a one-dimensional labeled array. It is like a single column of data.

Use a Series when you have one set of values and want to label each item.


```python
import pandas as pd

scores = pd.Series([85, 90, 78, 92], index=['Math', 'Science', 'English', 'History'])
print(scores)
print("\nValue for Math:", scores['Math'])
print("Average:", scores.mean())
```

## Key points
- A Series has values and an index.
- It is useful for one-dimensional data.
- You can create it from a list, dictionary, or NumPy array.
- Access values by label with `series['label']` or by position with `series.iloc[]`.
