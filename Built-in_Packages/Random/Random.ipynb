# Random Package

```python
import random
```

```python
# Get Random Number from 0-1

num1 = random.random()
print(num1)

```

```python
# Random Choice

names = ["Rohit", "Rahul","Akshay","Parth"]

choice = random.choice(names)
print(choice)

# random sample
samples = random.sample(names,2)
print(samples)
```

```python
# random in range
num3 = random.randrange(10,100)
print(num3)
```
