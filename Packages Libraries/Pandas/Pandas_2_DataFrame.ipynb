# Pandas DataFrame
A DataFrame is a two-dimensional table. It stores data in rows and columns, like a spreadsheet.

It is the most commonly used pandas object for real-world data analysis.


```python
import pandas as pd

students = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Paris', 'London']
}

df = pd.DataFrame(students)
print(df)
print("\nShape:", df.shape)
print("Columns:", list(df.columns))
```

## Accessing data
You can select a column, a row, or filter rows based on a condition.


```python
import pandas as pd

students = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Paris', 'London']
}
df = pd.DataFrame(students)

print(df['Name']) # Column Name can be accessed using the column name as a string in square brackets.

print(df.City) # Column City can also be accessed using the dot notation.

print(df[df['Age'] > 25]) # Rows where Age is greater than 25 can be accessed using boolean indexing.

print("First student: " + df['Name'][0]) # The first element of the Name column can be accessed using the index 0.
```
