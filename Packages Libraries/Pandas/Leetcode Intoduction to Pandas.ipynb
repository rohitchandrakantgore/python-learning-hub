# 2877 - Create DataFrame from a List

# 2880 - Select Data

```python
import pandas as pd

# Select rows where student_age == 21 and keep only two columns
students = pd.DataFrame({
    'student_id': [1, 2, 3, 4],
    'student_name': ['Rohit', 'Rohit', 'Rohit', 'Rohit'],
    'student_age': [20, 21, 22, 23],
    'student_department': ['IT', 'HR', 'Finance', 'Marketing']
})

result = students.loc[students['student_age'] == 21, ['student_id', 'student_name']]
print(result)
```

# 2882 - Drop Duplicate Rows
- Remove duplicate values from the email column and keep only the first occurrence.

```python
import pandas as pd

# Remove duplicate emails and keep the first occurrence
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'customer_name': ['Rohit', 'Rohit', 'Rohit', 'Rohit'],
    'email': ['rohit@gmail.com', 'rohit@gmail.com', 'rohit@yahoo.com', 'rohit@domain.com']
})

print(customers.drop_duplicates(subset=['email']))
```

# 2883 - Drop Missing Data

```python
import pandas as pd

# Drop rows where the name is missing
students = pd.DataFrame({
    'student_id': [1, 2, 3, 4],
    'name': ['Rohit', None, 'Alice', None],
    'age': [20, 21, 22, 23]
})

print(students.dropna(subset=['name']))
```

# 2884 - Modify Columns

# 2885 - Rename Columns

```python
import pandas as pd

# Rename columns for better readability
students = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'first': ['Rohit', 'Alice', 'Bob', 'Charlie'],
    'last': ['Sharma', 'Smith', 'Johnson', 'Williams'],
    'age': [20, 21, 22, 23]
})

students.rename(columns={
    'id': 'student_id',
    'first': 'first_name',
    'last': 'last_name',
    'age': 'age_in_years'
}, inplace=True)
print(students)
```

# 2886 - Change Data Type

```python
import pandas as pd

# Convert float values to integers
students = pd.DataFrame({
    'student_id': [1, 2, 3, 4],
    'name': ['Rohit', None, 'Alice', None],
    'age': [20, 21, 22, 23],
    'grade': [73.5, 85.0, 92.0, 78.9]
})
students['grade'] = students['grade'].astype('int')
print(students)
```

# 2887 - Fill Missing Data

```python
import pandas as pd

# Fill missing values with 0
products = pd.DataFrame({
    'name': ['Laptop', 'Smartphone', 'Tablet', 'Headphones'],
    'quantity': [10, None, None, 30],
    'price': [1000, 800, 500, 200]
})
products['quantity'] = products['quantity'].fillna(0)
print(products)
```

# 2888 - Reshape Data: Concatenate

```python
import pandas as pd

# Concatenate two DataFrames vertically
students_1 = pd.DataFrame({
    'student_id': [1, 2, 3],
    'name': ['Rohit', 'Alice', 'Bob'],
    'age': [20, 21, 22]
})
students_2 = pd.DataFrame({
    'student_id': [4, 5, 6],
    'name': ['Charlie', 'David', 'Eva'],
    'age': [23, 24, 25]
})

result = pd.concat([students_1, students_2], ignore_index=True)
print(result)
```

# 2889 - Reshape Data: Pivot

```python
import pandas as pd

# Pivot the data so each city becomes a column; pandas may sort the column labels
weather = pd.DataFrame({
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston'],
    'temperature': [30, 25, 20, 35],
    'month': ['January', 'February', 'March', 'April']
})

pivoted_weather = weather.pivot(index='month', columns='city', values='temperature')
print(pivoted_weather)
```

# 2890 - Reshape Data: Melt

```python
import pandas as pd

report = pd.DataFrame({
    'product': ['Laptop', 'Smartphone', 'Tablet', 'Headphones'],
    'quarter_1': [1000, 800, 500, 200],
    'quarter_2': [1200, 900, 600, 250],
    'quarter_3': [1100, 850, 550, 220],
    'quarter_4': [1300, 950, 650, 300]
})

# Convert wide data to long format: one row per product-quarter pair
reshaped_report = report.melt(
    id_vars=['product'],
    value_vars=['quarter_1', 'quarter_2', 'quarter_3', 'quarter_4'],
    var_name='quarter',
    value_name='sales'
)
print(reshaped_report)
```

# 2891 - Method Chaining

```python
import pandas as pd

animals = pd.DataFrame({
    'name': ['husky', 'tommy', 'bob', 'jonathan'],
    'species': ['dog', 'cat', 'dog', 'cat'],
    'age': [3, 5, 2, 4],
    'weight': [300, 100, 250, 50]
})

# Filter animals heavier than 100 kg, sort by weight descending, and reset the index
result = (
    animals.query('weight > 100')
    .sort_values('weight', ascending=False)
    .reset_index(drop=True)
)
print(result)
```
