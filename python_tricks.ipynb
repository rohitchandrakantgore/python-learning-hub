```python
# Different ways to test multiple
# flags at once in Python
x, y, z = 0, 1, 0

if x == 1 or y == 1 or z == 1:
    print('passed')

if 1 in (x, y, z):
    print('passed')

# These only test for truthiness:
if x or y or z:
    print('passed')

if any((x, y, z)):
    print('passed')
```

```python
# True False in python

values = [None, 0, "", False, [],{},(),"0",True]
for v in values:
    if v:
        print("True ",v)
    else:
        print("False ",v)

```
