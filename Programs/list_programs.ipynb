```python
# Find Duplicate Elements
my_list = [1,2,3,4,2,5,6,1]
duplicates = []
for i in my_list:
    if my_list.count(i) > 1 and i not in duplicates:
        duplicates.append(i)
print(duplicates)  # Output: [1, 2]
```

```python
# Flatten Nested List
my_list = [1, [2, [3, 4], 5], 6]
flattened = []
def flatten_list(nested_list):
    for item in nested_list:
        if isinstance(item, list):
            flatten_list(item)
        else:
            flattened.append(item)
    return flattened

print(flatten_list(my_list)) 
```

```python
# Find common elements with frequency

list1 = [1, 2, 3, 4, 5, 3, 4, 5, 5]
list2 = [3, 4, 5]

result = []
for each in list1:
    if each in list2:
        result.append(each)
print(result)

frequency ={}
for ch in list1:
    if ch in list2:
        frequency[ch]=  frequency.get(ch,0)+1


```
