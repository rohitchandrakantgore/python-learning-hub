```python
dict_input = {
    "home": [
        {
            "documents": [
                {
                    "pictures": [
                        "abc.jpg",
                        "def.png",
                        "abc.pdf"
                    ]
                },
                "def.pdf"
            ]
        },
        {
            "files": [
                "abc.pdf",
                "def.csv"
            ]
        },
        "result.pdf"
    ]
}


# home/documents/pictures/abc.jpg
# home/documents/pictures/def.png
# home/documents/pictures/abc.pdf
# home/documents/def.pdf
# home/files/abc.pdf
# home/files/def.csv
# home/result.pdf

paths = []

def dict_path(dict_input, current_path=""):
    if isinstance(dict_input,dict):
        for key,value in dict_input.items():
            new_path = f"{current_path}/{key}" if current_path else key
            dict_path(value, new_path)
    elif isinstance(dict_input,list):
         for each in dict_input:
            dict_path(each, current_path)
    else:
        paths.append(f"{current_path}/{dict_input}")
    return paths


result = dict_path(dict_input)
for each in result:
    print(each)
```
