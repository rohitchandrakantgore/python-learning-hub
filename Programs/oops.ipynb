```python
# Instance method, class method, and static method

class Employee:
    company = "Google"

    def __init__(self, name):
        self.name = name

    def instance_method(self): # Typical use: Working with individual employee data.
        # Receives self, which refers to the current object.
        # Can access instance attributes such as self.name.
        # Can also access class attributes through the instance.
        return self.name

    @classmethod
    def class_method(cls): # Typical use: Managing data shared by the class.
        # Receives cls, which refers to the class.
        # Can access or modify class attributes.
        # Often used for alternative constructors.
        return cls.company

    @staticmethod
    def static_method(): # Typical use: Helper functions related to employees, such as validating an employee ID.
        # Receives neither self nor cls automatically.
        # Does not automatically access instance or class state, but it can access class attributes by explicitly referencing the class.
        # Usually performs a utility operation related to the class.
        return "Employee Class"

emp = Employee("Rohit")
print(emp.instance_method())
print(emp.class_method())
print(emp.static_method())
```

| Feature | Instance method | Class method | Static method |
|---|---|---|---|
| Decorator | None | `@classmethod` | `@staticmethod` |
| First parameter | `self` | `cls` | No required special parameter |
| Receives | The current instance | The class | Neither instance nor class automatically |
| Accesses instance data | Yes, through `self` | No, not directly | No, not directly |
| Accesses class data | Yes | Yes, through `cls` | Only by explicitly referencing the class |
| Common use | Work with an individual object | Work with class-level data or create objects | Related utility functions |

```python
# Inheritance
# Inheritance is one of the four pillars of OOP. A child class can reuse methods and attributes from a parent class.

class Animal:
    def speak(self):
        return "Animal makes sound"

class Dog(Animal):
    def speak(self):
        #overrides the parent method
        return "Dog Barks"

class Cat(Animal):
    pass

dog = Dog()
cat = Cat()

print(dog.speak())
print(cat.speak())
```

- Inheritance: A child class reuses functionality from a parent class.
- Method overriding: A child class provides its own implementation of an inherited method.
- pass: A placeholder that does nothing.

# Polymorphism
-   Polymorphism means the same method name or interface can produce different behavior depending on the object.

```python
class Dog:
    def speak(self):
        return "Bark"

class Cat:
    def speak(self):
        return "Meow"

class Cow:
    def speak(self):
        return "Moo"

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.speak())
```

This is polymorphism: the same method call, animal.speak(), produces different behavior depending on the object's type.
Interview tip: Python supports duck typing. 
    If an object provides the required method, you can often use it without requiring all objects to inherit from the same parent class.

# Encapsulation
Encapsulation involves bundling data and methods together and controlling how internal state is accessed or modified.

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
        self._amount = balance

    def deposite(self,balance):
        self.__balance += balance

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposite(2000)

print(account.get_balance())
# print(account.__balance)
#This raises an AttributeError, because Python name-mangles the attribute __balance inside the class.
# Conceptually, Python changes its internal name to something like _BankAccount__balance
print(account._amount) # this will print
```

Is __balance truly private?
Not exactly. This is an important Python interview distinction:
- balance — public attribute.
- _balance — conventionally protected/internal; Python doesn't enforce access restrictions.
- __balance — triggers name mangling, which helps avoid accidental access and name collisions, but does not make the attribute strictly private.

Encapsulation bundles data and methods within a class and provides controlled access to internal state through methods or properties. Python uses naming conventions and name mangling rather than strict private-member enforcement.

# Abstraction
Abstraction hides implementation details and exposes only the required interface. Python supports this through the abc module.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass

class CreditCardPayment(Payment):
    def pay(self,amount):
        return f"Paid {amount} using credit card"

class UpiPayment(Payment):
    pass

payment1 = CreditCardPayment() # CreditCardPayment implements the abstract pay() method.
payment2 = UpiPayment()  # UpiPayment inherits from Payment but doesn't implement pay().
# Python raises the error here, so execution stops before reaching the print() statement.
print(payment1.pay(1000)) # If you remove the invalid payment2 creation, then it will return 
```

- Abstraction hides implementation details and defines a common interface. 
- In Python, ABC and @abstractmethod allow you to require subclasses to implement particular methods before their objects can be instantiated.
- For example, every payment type must implement pay(), but each payment type can implement the payment process differently.

# super() and inheritance

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def display(self):
        print(f"Animal: {self.name}")

class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name) # It calls the parent class's initializer.
        self.breed = breed

    def display(self):
        super().display() # It calls the parent class's display() implementation.
        print(f"Breed: {self.breed}")

dog = Dog("Buddy", "Labrador")
dog.display()
    
```

super() is useful for reusing parent-class behavior without hardcoding the parent class name. In multiple inheritance, it also follows Python's method resolution order (MRO).

# Method Resolution Order (MRO)

```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")
        super().show()

class C(A):
    def show(self):
        print("C")
        super().show()

class D(B,C):
    def show(self):
        print("D")
        super().show()

obj = D()
obj.show()

# When you call obj.show():
# 1. D.show() prints D, then calls super().show().
# 2. B.show() prints B, then calls super().show().
# 3. C.show() prints C, then calls super().show().
# 4. A.show() prints A.

# uper() doesn't simply call the direct parent. It calls the next class in the MRO. This is especially important in multiple inheritance.
```

# Constructor inheritance

```python
class Parent:
    def __init__(self):
        print("Parent Constructor")

class Child(Parent):
    def __init__(self):
        print("Child Constructor")
    

obj = Child()

class Child2(Parent):
    def __init__(self):
        super().__init__()
        print("Child2 constructor")

obj = Child2()
```

- If a child class does not define its own __init__(), it inherits the parent's constructor. 
- If it defines one, the parent constructor runs only if explicitly called, for example, using super().__init__()

# Composition vs. inheritance
| Inheritance | Composition | 
|---|---|
|Represents an is-a relationship |	Represents a has-a relationship |
|class Dog(Animal):	|self.engine = Engine()|
|Dog is an Animal	|Car has an Engine|

Composition helps build flexible systems by combining independent objects. For example, a Car can use different engine implementations without needing to inherit from Engine

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine() # Car creates and stores an Engine object as an instance attribute

    def start(self):
        self.engine.start()
        #The super() function is used to access the next class in the method resolution order, commonly when working with inheritance.
        print("Car started")

car = Car()
car.start()
```

Class variables and object counting

```python
class Employee:
    count = 0

    def __init__(self,name):
        self.name = name
        Employee.count+=1

emp1 = Employee("Amit")
emp2 = Employee("Priya")
emp3 = Employee("Rahul")

print(Employee.count)
print(emp1.count)

# - emp1 created → count = 1
# - emp2 created → count = 2
# - emp3 created → count = 3
# Employee.count accesses the class variable directly.
# emp1.count also works because Python looks for the attribute on the instance first. If it doesn't find it, Python looks for the attribute on the class, where count = 3 exists.
# Interview tip: Employee.count is generally clearer when you want to access a class-level counter.
```

Class variable vs. instance variable

```python
class Employee:
    count = 0
    def __init__(self,name):
        self.name = name
        Employee.count +=1

emp1 = Employee("Amit")
emp2 = Employee("Priya")

emp1.count = 100

print(Employee.count)
print(emp1.count)
print(emp2.count)

# Python creates an instance attribute named count on emp1. This shadows the class variable for that particular instance.
# Expression	Value	Reason
# Employee.count	2	Original class variable
# emp1.count	100	Instance attribute shadows the class variable
# emp2.count	2	Still reads the class variable
# Interview takeaway: Assigning to obj.attribute can create an instance attribute that shadows a class attribute. It doesn't modify the class attribute itself.
```

# @property in Python

 @property defines a getter.
- @price.setter defines a setter.
- Both let you use product.price as an attribute while running validation logic behind the scenes.

```python
class Product:
    def __init__(self,price):
        self._price = price

    @property  # This lets us access the method like an attribute
    def price(self):
        return self._price

    @price.setter
    def price(self,value):
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value

product = Product(100)
print(product.price) # calls the price() method automatically when you access product.price

product.price = 200
print(product.price)

product.price = -50
print(product.price)
```

# Dunder methods (__str__)
- Dunder methods are special methods in Python, such as _ _init_ _, _ _str_ _, and _ _len_ _

__str__() returns a string representation of an object

```python
class Employee:
    def __init__(self,name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"{self.name} earns {self.salary}"

emp = Employee("Amit", 60000)
print(emp)
print(str(emp))

# print(emp) calls str(emp) internally.
# str(emp) invokes the object's __str__() method.
# __str__() returns a readable string describing the employee.
```

__str__() is intended for a human-readable representation, while __repr__() is intended for a more developer-oriented representation.

__str__() provides a human-friendly representation of an object. It is used by print() and str().

__repr__() provides a developer-oriented representation, useful for debugging and inspecting objects. It is used by repr().

```python
# __str__() vs. __repr__()

class Employee:
    def __init__(self,name):
        self.name = name

    def __str__(self):
        return f"Employee Name: {self.name}"

    def __repr__(self):
        return f"Employee('{self.name}')"

emp = Employee("Amit")
print(emp)
print(repr(emp))
```

# Operator overloading

```python
class Book:
    def __init__(self,pages):
        self.pages = pages

    def __add__(self,other):
        return self.pages + other.pages

book1 = Book(100)
book2 = Book(200)
print(book1+book2) # internally book1.__add__(book2)

# This is called operator overloading: defining how an operator such as + behaves for your custom objects.
# Interview tip: Other common special methods include __sub__() for -, __mul__() for *, and __eq__() for ==.
```

# Object equality (== vs. is)

```python
class Employee:
    def __init__(self, name):
        self.name = name

emp1 = Employee("Amit") # Object A — name: Amit
emp2 = Employee("Amit") # Object B — name: Amit
emp3 = emp1  # Object A — same object as emp1

print(emp1 == emp2)
print(emp1 is emp2)
print(emp1 is emp3)
# - == checks equality of values or objects, according to the type's equality implementation.
# - is checks object identity.
# - emp3 = emp1 makes both variables refer to the same object; it doesn't copy the object.
```

# Shallow copy vs. shared reference

```python
class Employee:
    def __init__(self,name,skills):
        self.name = name
        self.skills= skills

emp1 = Employee("Amit",["Python","SQL"])
emp2 = emp1

emp2.skills.append("AWS")

print(emp1.skills)
print(emp2.skills)

#Assignment creates another reference, not a copy. 
# If you need a separate object, you can use copy.copy() for a shallow copy or
#  copy.deepcopy() for a deep copy, depending on whether nested mutable objects must also be independent.
```

```python
# Shallow copy
import copy

class Employee:
    def __init__(self,name,skills):
        self.name = name
        self.skills = skills

emp1 = Employee("Amit",["Python","SQL"])
emp2 = copy.copy(emp1)

emp2.name = "Rahul"
emp2.skills.append("AWS")

print(emp1.name)
print(emp2.name)
print(emp1.skills)
print(emp2.skills)
```

```python
# Deep Copy
import copy

class Employee:
    def __init__(self,name,skills):
        self.name = name
        self.skills = skills

emp1 = Employee("Amit",["Python","SQL"])
emp2 = copy.deepcopy(emp1)

emp2.skills.append("AWS")
print(emp1.skills)
print(emp2.skills)
```

# Instance method vs. class method

```python
class Employee:
    company = "OpenAI"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company
        #This changes the class variable, not an instance variable. 

emp1 = Employee("Amit")
emp2 = Employee("Priya")

emp1.change_company("Google") # invokes the class method with cls referring to the Employee class.

print(emp1.company)
print(emp2.company)
print(Employee.company)
#A class method can be called through either the class (Employee.change_company(...)) 
# or an instance (emp1.change_company(...)). In both cases, cls refers to the class.
```

# Class method vs. instance attribute
instance first, then class for ordinary instance attribute lookup, followed by further lookup mechanisms when applicable

```python
class Employee:
    company = "OpenAI"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

emp1 = Employee("Amit")
emp2 = Employee("Priya")

emp1.company = "Microsoft" # This creates an instance attribute on emp1. 
#It shadows the class variable for that instance only.
Employee.change_company("Google")
# The class method updates Employee.company to "Google". 
# It doesn't remove or overwrite emp1's instance attribute.

print(emp1.company)
print(emp2.company)
print(Employee.company)
```

# Polymorphism with a common interface

```python
class CreditCard:
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")

class UPI:
    def pay(self, amount):
        print(f"Paid {amount} using UPI")

class PayPal:
    def pay(self, amount):
        print(f"Paid {amount} using PayPal")

def process_payment(payment_method, amount):
    payment_method.pay(amount)

payments = [CreditCard(), UPI(), PayPal()]

for payment in payments:
    process_payment(payment, 500)
```

```python
#You need to create a Shape base class and two subclasses: Circle and Rectangle.
# Requirements:
# - Circle takes radius.
# - Rectangle takes width and height.
# - Both classes implement an area() method.
# - Use math.pi to calculate the circle's area.
# Your task: Write the Python code to create both classes, create one circle with radius 5 
# and one rectangle with width 4 and height 6, and print their areas.

import math 
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

cir = Circle(5)
rect = Rectangle(4,6)
print(cir.area())
print(rect.area())
```
