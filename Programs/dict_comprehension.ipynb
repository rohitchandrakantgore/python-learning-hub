# Dictionary comprehension uses this syntax:
    {key_expression: value_expression for item in iterable}

# With a condition:
    {key: value for item in iterable if condition}

# With if-else:
    {key: value_if_true if condition else value_if_false for item in iterable}

```python
# Number and square
nums = [1, 2, 3, 4, 5]
result = {x:x**2 for x in nums}
result
```

```python
# Filter even numbers and calculate their cubes
nums = [1, 2, 3, 4, 5, 6, 7, 8]

result = {x : x**3 for x in nums if x%2 == 0}
print(result)
```

```python
#Swap dictionary keys and values
data = {"a": 1, "b": 2, "c": 3}
result = {value:key for key,value in data.items()}
print(result)
```

```python
# Convert names to uppercase
names = ["amit", "priya", "rahul"]
result = {x:x.upper() for x in names}
print(result) 
```

```python
# Conditional values — Even or Odd
nums = [1, 2, 3, 4, 5, 6]
result = {x:"Odd" if x%2!=0 else "Even" for x in nums}
print(result)
```

```python
# Filter dictionary values
scores = {
    "Amit": 85,
    "Priya": 45,
    "Rahul": 72,
    "Neha": 35
}
result = {key:value for key,value in scores.items() if value >= 60}
print(result)
```

```python
# Word length mapping
words = ["python", "django", "api", "fastapi"]
result = {x:len(x) for x in words}
print(result) 
```

```python
# Extract active users — Backend interview question
users = [
    {"name": "Amit", "active": True},
    {"name": "Priya", "active": False},
    {"name": "Rahul", "active": True},
    {"name": "Neha", "active": False}
]
result = {x["name"]:x["active"] for x in users if x["active"]==True}
print(result)

# improvement
result = {x["name"]:x["active"] for x in users if x["active"]}
print(result)
```

```python
# Group numbers by even and odd
nums = [1, 2, 3, 4, 5, 6, 7, 8]
result = {
            "Even":[ x for x in nums if x%2 == 0] ,
            "Odd":[ x for x in nums if x%2 != 0 ]
        }
result
```

```python
# Create a dictionary from two lists
keys = ["name", "age", "city"]
values = ["Rahul", 28, "Pune"]
result = {key:value for key,value in zip(keys,values)}
print(result)
```

```python
# Transform dictionary values
prices = {
    "apple": 100,
    "banana": 40,
    "orange": 80
}
result = {
            key : value + value*0.10 
            for key,value in prices.items()}
print(result)

# A cleaner alternative
result = {key: value * 1.10 for key, value in prices.items()}
print(result)
```

```python
# Filter and transform a dictionary
employees = {
    "Amit": 45000,
    "Priya": 65000,
    "Rahul": 55000,
    "Neha": 80000,
    "Vijay": 35000
}
result = {
    key:value*1.10
    for key,value in employees.items()
    if value >= 55000 
}
print(result)
```

```python
# Group words by their first letter
words = ["apple", "ant", "banana", "ball", "cat", "car"]
result = {
    letter:[word for word in words if word.startswith(letter)]
    for letter in {word[0] for word in words}
}
print(result)
```

```python
#Count character frequencies
text = "programming"

result = {x: text.count(x) for x in text}

print(result)

# improved version - for not to traverse repeated characters
result = {
    x: text.count(x)
    for x in dict.fromkeys(text)
}
print(result)
```

```python
# Group numbers by their remainder
nums = [10, 11, 12, 13, 14, 15, 16, 17]

result = {
    0: [x for x in nums if x%2 == 0],
    1 : [x for x in nums if x%2 != 0]
}
print(result)

# A true dictionary comprehension solution is:
result = {
    remainder: [x for x in nums if x % 2 == remainder]
    for remainder in (0, 1)
}
print(result)
```

```python
# Advanced dictionary comprehension
#Create a dictionary containing only IT employees, with each employee's name as the key and salary as the value.
employees = [
    {"name": "Amit", "department": "IT", "salary": 60000},
    {"name": "Priya", "department": "HR", "salary": 50000},
    {"name": "Rahul", "department": "IT", "salary": 75000},
    {"name": "Neha", "department": "HR", "salary": 65000}
]

result = {
    x["name"]:x["salary"] for x in employees
    if x["department"] == "IT"
}
result
```

```python
# Group employees by department
employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Priya", "department": "HR"},
    {"name": "Rahul", "department": "IT"},
    {"name": "Neha", "department": "HR"},
    {"name": "Vijay", "department": "Finance"}
]

result = {
    dept:[
        emp["name"]
        for emp in employees
        if emp["department"] == dept
    ]
    for dept in {emp['department'] for emp in employees}
}
print(result)
```

```python
# Count employees by department
employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Priya", "department": "HR"},
    {"name": "Rahul", "department": "IT"},
    {"name": "Neha", "department": "HR"},
    {"name": "Vijay", "department": "Finance"}
]
result = {
    dept : sum(
        1
        for emp in employees 
        if emp['department']==dept)
    for dept in {emp['department'] for emp in employees}
}
result
```

```python
# Invert a dictionary with duplicate values
data = {
    "Amit": "IT",
    "Priya": "HR",
    "Rahul": "IT",
    "Neha": "HR",
    "Vijay": "Finance"
}

result = {
    dept:[
        name for name in data.keys()
        if data[name]== dept
    ]
    for dept in set(data.values())
}
print(result)
```

```python
#Filter and transform a nested dictionary 
# Create a dictionary containing only students who scored 60 or more, with their names as keys and their marks increased by 5 as values.
students = {
    "Amit": {"marks": 85, "age": 22},
    "Priya": {"marks": 45, "age": 21},
    "Rahul": {"marks": 92, "age": 23},
    "Neha": {"marks": 60, "age": 20}
}

result = {
    name: students[name]['marks'] + 5
    for name in students
    if students[name]["marks"] >= 60
}
print(result)

# Interview tip: Use .items() for cleaner code
# You can avoid looking up students[name] repeatedly by unpacking the name and details directly:
result = {
    name: details["marks"] + 5
    for name, details in students.items()
    if details["marks"] >= 60
}
print(result)
```

```python
# Create a dictionary where each product maps to its total sales amount.
orders = [
    {"product": "Laptop", "price": 50000},
    {"product": "Mouse", "price": 500},
    {"product": "Laptop", "price": 60000},
    {"product": "Keyboard", "price": 1500},
    {"product": "Mouse", "price": 700}
]

result = {
    product: sum(
        order['price']
        for order in orders
        if product == order['product']
    )
    for product in {order['product'] for order in orders}
}
print(result)

# normale version with dict comprehension
result = {}
for order in orders:
    product = order['product']
    price = order['price']
    result[product] = result.get(product,0) + price

print(result)
```

```python
#Create a dictionary containing only employees who:
# - Work in the IT department.
# - Are active.
# - Earn at least 60000.
employees = {
    "Amit": {"department": "IT", "salary": 70000, "active": True},
    "Priya": {"department": "HR", "salary": 50000, "active": True},
    "Rahul": {"department": "IT", "salary": 45000, "active": False},
    "Neha": {"department": "IT", "salary": 80000, "active": True},
    "Vijay": {"department": "Finance", "salary": 90000, "active": True}
}

result = {
        name : details['salary']
        for name,details in employees.items()
        if details['department'] == "IT"
        and details['salary'] >= 60000
        and details['active']
    }
print(result)
```
