# Filtering: if comes after the for clause
[x for x in nums if x > 5]

# Transformation with a condition: if-else comes before the for clause
["High" if x > 5 else "Low" for x in nums]

```python
# list of numbers
result = [x for x in range(10)]
print(result)
```

```python
nums = [1,2,3,4,5,6,7,8,9]

# replace 5 with 1
result= [1 if x==5 else x for x in nums  ]
print(result)
```

```python
#Square every number in a list
nums = [1, 2, 3, 4, 5]
sqr_list = [x**2 for x in nums]
sqr_list
```

```python
#Extract even numbers
nums = [1, 2, 3, 4, 5, 6, 7, 8]
even_nums= [x for x in nums if x%2==0]
even_nums
```

```python
# Convert strings to uppercase
names = ["python", "django", "fastapi"]
upper_str = [x.upper() for x in names]
upper_str
```

```python
# Replace negative numbers with zero
nums = [10, -5, 20, -3, 0, 8]
nums_res = [x if x > 0 else 0  for x in nums ]
nums_res
```

```python
#Extract numbers greater than 10 and multiply them by 2
nums = [5, 12, 8, 15, 20, 3]
result = [x*2 for x in nums if x>10]
result
```

```python
# Replace 5 with 100 and keep other values unchanged
nums = [1, 5, 3, 5, 7, 5]
result = [100 if x==5 else x for x in nums]
result
```

```python
# Label numbers as even or odd
nums = [1, 2, 3, 4, 5]
result= ['even' if x%2==0 else 'odd' for x in nums]
print(result)
```

```python
#Extract words containing more than 4 characters
words = ["cat", "python", "java", "django", "go", "fastapi"]
result= [x for x in words if len(x)>4]
result
```

```python
#Replace even numbers with their squares and odd numbers with their cubes
nums = [1, 2, 3, 4, 5]
result = [x**2 if x%2==0 else x**3 for x in nums]
result
```

```python
# Remove empty strings and convert the remaining strings to lowercase
words = ["PYTHON", "", "DJANGO", " ", "FASTAPI", ""]
result = [x.lower() for x in words if len(x)>0]
result
```

```python
# Flatten a nested list
matrix = [[1, 2], [3, 4], [5, 6]]
result = [x for row in matrix for x in row]
print(result)

# Normal Version
result = []
for row in matrix:
    for x in row:
        result.append(x)
print(result)
```

```python
# Generate all pairs from two lists
a = [1, 2]
b = ["A", "B"]
result = [(x,y) for x in a for y in b]
print(result)

# normal version
result = []
for x in a:
    for y in b:
        result.append((x, y))
print(result)
```

```python
#Find common elements between two lists
a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]

result = [x for x in a if x in b]
print(result)

#For better performance with large lists, use a set:
b_set = set(b)

result = [x for x in a if x in b_set]

print(result)
```

```python
# Extract vowels from a string
text = "programming"

result = [x for x in text if x in ('a','e','i','o','u') ]
print(result)
```

```python
# Create a list of tuples containing a number and its square
nums = [1, 2, 3, 4]
result = [(x,x**2) for x in nums]
print(result)
```

```python
#Flatten a 2D matrix and keep only even numbers

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

result = [x for row in matrix for x in row if x%2==0]
print(result)
```

```python
# Find all numbers divisible by both 3 and 5
nums = list(range(1, 51))
result = [x for x in nums if x%3==0 and x%5==0]
print(result)
```

```python
# Remove duplicates while preserving order
nums = [1, 2, 2, 3, 1, 4, 3, 5]
result = [x for x in set(nums)]
print(result)
```

```python
# Generate a multiplication table
n = 5
result = [x*n for x in range(1,11)]
print(result)
```

```python
# Extract names of active users
users = [
    {"name": "Amit", "active": True},
    {"name": "Priya", "active": False},
    {"name": "Rahul", "active": True},
    {"name": "Neha", "active": False}
]

result = [x["name"] for x in users if x["active"]==True]
print(result)
```
