# Palindrome String

**Problem:** Check whether a string reads the same forward and backward, ignoring spaces and letter case.


```python
def is_palindrome(text):
    """Return True if the normalized text is a palindrome."""
    normalized = ''.join(character.lower() for character in text if character.isalnum())
    return normalized == normalized[::-1]


print(is_palindrome("madam"))
print(is_palindrome("nurses run"))
print(is_palindrome("Python"))
```
