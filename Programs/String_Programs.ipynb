```python
#Check if Two Strings are Anagrams
str1,str2 = "listen", "silent"
# method 1
def is_anagram_1(str1, str2):
    return sorted(str1.lower()) == sorted(str2.lower())

# method 2
from collections import Counter
def is_anagram_2(str1, str2):
    return Counter(str1.lower()) == Counter(str2.lower())

print(is_anagram_1(str1, str2))  # Output: True
print(is_anagram_2("hello", "world"))  # Output: False
```

```python
#Group Words by Length
words = ["apple", "banana", "cherry", "date", "elderberry"]
grouped_words = {}
for word in words:
    length = len(word)
    if length not in grouped_words:
        grouped_words[length] = []
    grouped_words[length].append(word)
print(grouped_words)  # Output: {5: ['apple'], 6: ['banana'], 7: ['cherry'], 4: ['date'], 10: ['elderberry']}
```

```python
# Longest Substring Without Repeating Characters
s = "abcabcbb"
def longest_unique_substring(s):
    seen = {}
    left = 0
    max_length = 0

    for right in range(len(s)):
        char = s[right]

        if char in seen and seen[char] >= left:
            left = seen[char] + 1

        seen[char] = right
        max_length = max(max_length, right - left + 1)

    return max_length

print(longest_unique_substring("abcabcbb"))
```

```python
# Mini Password Validator
# Rules:

# 8+ chars
# uppercase
# lowercase
# number
# special char

import re

def validate_password(password):
    if len(password) < 8:
        return False

    patterns = [
        r'[A-Z]',  # Uppercase letter
        r'[a-z]',  # Lowercase letter
        r'[0-9]',  # Digit
        r'[\W_]'   # Special character (non-word character)
    ]

    for pattern in patterns:
        if not re.search(pattern, password):
            return False    
    return True

print(validate_password("Password123!"))  # Output: True
print(validate_password("password"))      # Output: False
```
