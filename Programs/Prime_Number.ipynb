# Prime Number

**Problem:** Determine whether a positive integer is prime. A prime number has exactly two positive factors: 1 and itself.


```python
import math


def is_prime(number):
    """Return True when number is prime; otherwise, return False."""
    if number < 2:
        return False

    for divisor in range(2, math.isqrt(number) + 1):
        if number % divisor == 0:
            return False
    return True


for value in (1, 2, 11, 15, 29):
    print(f"{value}: {is_prime(value)}")
```
