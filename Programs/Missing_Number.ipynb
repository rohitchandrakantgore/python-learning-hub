# Missing Number

**Problem:** Given distinct integers containing every number from 1 to n except one, find the missing integer.


```python
nums = [1, 2, 4, 5, 6]


def find_missing_number(nums):
    """Find the missing value using expected and actual sums."""
    n = len(nums) + 1
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum


print(find_missing_number(nums))
```
