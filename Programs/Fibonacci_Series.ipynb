# Fibonacci Series

**Problem:** Generate the first n numbers in the Fibonacci sequence: 0, 1, 1, 2, 3, 5, and so on.


```python
def fibonacci_generator(n):
    """Yield the first n Fibonacci numbers."""
    if n < 0:
        raise ValueError("n must be non-negative")

    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b


for num in fibonacci_generator(10):
    print(num)
```
