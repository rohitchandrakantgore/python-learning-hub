# Arrays
-   collection of elements stored at contiguous memory locations
-   store multiple items of same datatype
-   syntax:
    - array_name = array('type_code',[elements])


- type codes - 
    - 'i'   - int
    - 'b'   - signed char
    - 'f'   - float
    - 'd'   - double

```python
# Creating of Array

import array 

new_array = array.array('i',[1,2,3,4,5])
print(new_array)

# accessing array elements
print("first_element: ", new_array[0])
# accessing_last_element
print("last_element: ",new_array[-1])

# reverse array
new_array.reverse()
print(new_array)
```
